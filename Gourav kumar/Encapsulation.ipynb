# Encapsulatin means wrapping data and methods together inside a class and controlling how that data can be accessed.

```python
# example:
balance = 10000
balance = -50000
```

# public

# public members can be accessed from anywhere. In python, normally we simply write the variable name.

```python
class Student:
    def __init__(self):
        self.name = "Rahul"    # public

s1 = Student()

print(s1.name)
```

```python
s1.name = "Ankit"

print(s1.name)
```

# private

# Private members are written with two underscores__.

```python
class Bank:
    def __init__(self):
        self.__balance = 10000

b = Bank()

print(b.__balance)
```

```python
class Bank:
    def __init__(self):
        self._balance = 10000

b = Bank()

print(b._balance)
```

# we can acess the data by adding other user define function in it

```python
class Bank:
    def __init__(self):
        self.__balance = 10000

    def show_balance(self):
        print(self.__balance)

b = Bank()

b.show_balance()
```

# Protected

# Protected behaves more like public because you technically can access it from outside, but the _ tells programmers:"This is intended for internal/subclass use. 

```python
class Student:
    def __init__(self):
        self._marks = 80

s1 = Student()

print(s1._marks)
```

# Best easy example

```python
class Student:
    def __init__(self):
        self.name = "Rahul"       # public
        self._age = 20             # protected
        self.__marks = 90         # private


s = Student()

print(s.name)
print(s._age)
print(s.__marks)
        
```
