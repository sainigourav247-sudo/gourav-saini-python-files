# Easy

```python
double = lambda x: x*2
print(double(6))
```

```python
square = lambda x:x**2
print(square(4))
```

```python
add = lambda x, y: x+y
print(add(5,3))
```

# medium

```python
# cube
cube = lambda x: x**3
print(cube(3))
```

```python
# even or odd
is_even = lambda x: "even" if x % 2 == 0 else "odd"
print(is_even(4))
print(is_even(7))
```

```python
larger = lambda x,y:x if x > y else y
print(larger(10,5))
```

```python
smaller = lambda x, y: x if x <y else y
print(smaller(10,5))
```

# hard

```python
check_sign = lambda x: "positive" if x > 0 else "negative" if x < 0 else "zero"
print(check_sign(5))
print(check_sign(-3))
print(check_sign(0))
```

```python
numbers = [1,2,3,4,5]
doubled = list(map(lambda x: x* 2, numbers))
print(doubled)
```

```python
numbers = [10,15,20,25,30,35]
evens= list(filter(lambda x: x % 2 == 0, numbers))
print(evens)
```
