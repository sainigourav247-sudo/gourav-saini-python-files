# 9. Fizz buzz question

```python
for i in range(1,101):
    if i%3==0 and i%5==0:
        print("fizzbuzz")
    elif i%5==0:
        print("buzz")
    elif i%3==0:
        print("fizz")
    else:
        print(i)
```

# 1. write a program using a for loop to count the numbers from 1 to 20.

```python
count=0
for i in range (3,22):
    count=count+1
print(count)
```

# 2.write a program using a for loop to count how many even numbers are between 1 to 50.

```python
count=0
for i in range (1,53):
    if i%2==0:
        count=count+1
print("even number","=",count)
```

# 3.write a program using a for loop to find the total of numbers from 1 to 10.

```python
total=0
for i in range (1,11):
    total= total+i
print("total numbers =",total)
```

# 4. write a program using a for loop to find the sum of all odd numbers from 1 to 20.

```python
c=0
for i in range (1,21):
    if i%2!=0:
        c=c+1
print(c)
```

# 5. write a program using a for loop to find the average of numbers from 1 to 10.

```python
total=0
num=0
for i in range (1,11):
    total= total+i
    num=num+1
    average=total/num
print("average =",average)
```

# 6.write a program using a for loop to find the average of the numbers 10,20,30,40, and 50.

```python
number=10,20,30,40,50
total=0
num=0
for i in number:
    total= total+i
    num=num+1
    average=total/num
print("average =",average)
```
