# 1 positional arguments

```python
def student(name,age):
    print("name:",name)
    print("age:",age)

student("rahul",20)
```

```python
def student(age,name):
    print("age:",age)
    print("name:",name)
```

```python
student("ram",55)
```

# 2 keyword arguments

```python
def student(name="ram",age=66):
    print(name,age)

student()
```

# 3 positional+ keywords 

```python
def student(name,age,city):
    print(name,age,city)

student("rahul",age=20,city="amritsar")
```

# *args

```python
def add(*hi):
    print(hi)

add(10,20,30,40)
```

```python
def add(*a): 
    print(a)
    
add(10,20,30,40)
```

```python
def add (*args):
    total=0

    for number in args:
        total+=number

    print(total)

add(30,69,45)
```

# **kwargs

```python
def student(**kwargs):
    print(kwargs)

student(name="rahuk",age=20,city="amritsar")
```
