```python
score=int(input("enter your number:"))
if score >=90:
    print("grade: a")
elif score>=80:
    print("grade:b")
elif score>=70:
    print("grade:c")
elif score>60:
    print("grade:d")
else:
    print("grade:f")
```

```python
light= input("enter traffic light color")
if light=="red":
    print("stop")
elif light=="yellow":
    print("prepare to stop")
elif light=="green":
    print("go")
else:
    print("invalid color")
```

```python
num= int(input("enter your number:"))
if (num>0) and (num%2==0):
    print("positive even")
elif (num>0) and (num%2!=0):
    print("negative even")
elif (num<0) and (num%2!=0):
    print("negative odd")
else:
    print("zero")
```

```python
age= int(input("enter your age"))
if age <=13:
    print("child")
elif age <=19:
    print("teenager")
elif age <=59:
    print("adult")
else:
    print("senior")
```
