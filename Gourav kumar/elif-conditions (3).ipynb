# elif - conditions

# 1

```python
score= int(input("Enter your Number: "))
if score >=90:
    print("Grade: A")
elif score >=80:
    print("Grade: B")
elif score >=70:
    print("Grade: C")
elif score >=60:
    print("Grade: D")
else:
    print("Grade: F")
```

# 2

```python
light= input("Enter traffic light color: ")
if light=="red":
    print("Stop")
elif light=="yellow":
    print("Prepare to Go")
elif light=="green":
    print("Go")
else:
    print("Invalid color")
```

# 3

```python
num= int(input("Enter a number: "))
if (num >0) and (num%2==0):
    print("Positive Even")
elif (num >0) and (num%2!=0):
    print("Positive odd")
elif (num <0) and (num%2==0):
    print("Negative even")
elif (num <0) and (num%2!=0):
    print("Negative odd")
else:
    print("Zero")
```

# 4

```python
age= int(input("Enter your age: "))
if age <=13:
    print("Child")
elif age <=19:
    print("Teenager")
elif age <=59:
    print("Adult")
else:
    print("Senior")
```

# 5

```python
amount = float(input("Enter purchase amount: $"))
if amount < 20:
    discount = 0
elif amount <= 50:
    discount = 5
elif amount <= 100:
    discount = 10
else:
    discount = 15
discount_amount = amount * discount / 100
final_price = amount - discount_amount
print("Discount:", discount, "%")
print("discount_amount:$",discount_amount)
print("Final Price: $", final_price)
```
