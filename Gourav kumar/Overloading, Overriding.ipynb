# Overloading  ,  Overriding

# Meaning: A child class provides its own version of a method that already exists in the parent class.

```python
class Father:  
    def work(self):
        print("Father is working")


class Son(Father):
    def work(self):
        print("Son is working")


s = Son()
s.work()
```

```python
class Father(Son):  
    def work(self):
        print("Father is working")


class Son:
    def work(self):
        print("Son is working")


s = Father()
s.work()
```

# Method Overloading

# Meaning: Using the same method name with different numbers/types of arguments.

```python
class Calculator:

    def add(self, a, b):
        return a + b

    def add(self, a, b, c):
        return a + b + c
```

```python
c= Calculator()
```

```python
c.add(10,20,30)
```

# The second add() replaces the first one.

```python
class Calculator:

    def add(self, a, b, c=0):
        return a + b + c 


c = Calculator()

print(c.add(10,20))
print(c.add(10,20,30))
```

# super function

```python
class Father:
    def work(self):
        print("Father is working")

class Son(Father):
    def work(self):
        super().work()
        print("Son is working")


s = Son()
s.work()
```

# Super() with init()

```python
class Father:
    def __init__(self):
        print("Father constructor")

class Son(Father):
    def __init__(self):
        super().__init__()
        print("Son constructor")


s = Son()
```
